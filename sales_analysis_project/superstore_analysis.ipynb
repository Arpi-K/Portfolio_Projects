```python
#which category has the highest total sales and total proofit
category_perf = df_clean.groupby("category").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum")
).sort_values(by="Total_Sales", ascending=False)
category_perf
```

```python
#top 10 countries by sales
top_countries=df_clean.groupby("country")["sales"].sum().nlargest(10).reset_index()
top_countries
```

```python
#which category has the highest sales?
import matplotlib.pyplot as plt
category_perf.plot(kind='bar',figsize=(8,6))
plt.ticklabel_format(style='plain',axis='y')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales/Profit")
plt.show()
```

```python
#top 10 products by sales
top_products=df_clean.groupby("product_name")["sales"].sum().nlargest(10).reset_index()
top_products
```

```python
#top 10 customers by sales
top_customers=df_clean.groupby("customer_name")["sales"].sum().nlargest(10).reset_index()
top_customers
```

```python
#sales by market
sales_by_market=df_clean.groupby("market",as_index=False)["sales"].sum().sort_values(by="sales",ascending=False).reset_index()
sales_by_market
```

```python
#profit by segment
profit_by_seg=df_clean.groupby("segment",as_index=False)["profit"].sum().sort_values(by="profit",ascending=False).reset_index()
profit_by_seg
```

```python
#shipping mode usage
ship_mode_usage=df_clean['ship_mode'].value_counts().reset_index(name="order count")
ship_mode_usage
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df_clean['discount'],df_clean['profit'],alpha=0.4)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
